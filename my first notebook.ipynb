### calculating Letter Grade

- 2 or under D
- 3 or under C
- 3.6 or under B
- anything above 3.6 A

```python
student_gpa = gpa
if student_gpa <=2:
    letter_grade= "D"
elif student_gpa <=3:
    letter_grade= "C"
elif student_gpa <=3.6:
    letter_grade = "B"
else:
    letter_grade = "A"

print(f"{name}'s letter grade is {letter_grade}")
```

### 
List Comprehension in Python

```python
top_3=[]
for s in sports:
    top_3.append(s[:3])
print(top_3)
```

```python
# let's use list comprehension to make the new sports list
top3_v2 = [s[:3] for s in sports]

print(top3_v2)
```

```python
first_last=[s[:1]+s[-1:] for s in sports]
print(first_last)
```
